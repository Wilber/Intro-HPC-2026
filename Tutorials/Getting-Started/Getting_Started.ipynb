In this example we'll use a Jupyter notebook to: 
- inspect data

- make a plot

### 2. Import a library

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### 3. Create some data

```python
#create the data
data = { 
    "Student": ["Alex", "Jordan", "Sam", "Taylor", "Morgan"], 
    "StudyHours": [2, 4, 5, 7, 8], 
    "QuizScore": [68, 76, 82, 91, 95] 
}
```

```python
df = pd.DataFrame(data)
```

### 4. Explore the data with:
- `df.head()`
- `df.tail()`
- `df.describe()`

```python
df.head()
```

### 5. Change something

```python
df.head(3)
```

### 6. Calculate something

```python
average_score = df["QuizScore"].mean() 

print(average_score)
```

### 7. Create a visualization

```python
plt.scatter(df["StudyHours"], df["QuizScore"])

plt.xlabel("Hours Studied") 
plt.ylabel("Quiz Score") 

plt.title("Study Hours vs. Quiz Score");
```
